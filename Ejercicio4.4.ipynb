```python
class Vehiculo:
    """Clase base (padre) que define la interfaz común."""

    def __init__(self, placa: str, marca: str):
        self.placa = placa
        self.marca = marca

    def calcular_tarifa(self) -> float:
        """Método polimórfico que será sobrescrito por las clases hijas."""
        return 0.0

    def mostrar_detalle(self):
        """Muestra los datos básicos del vehículo."""
        print(f"Placa: {self.placa} | Marca: {self.marca}")


class Auto(Vehiculo):
    """Clase hija que representa un Automóvil."""

    def __init__(self, placa: str, marca: str, es_suv: bool):
        super().__init__(placa, marca)
        self.es_suv = es_suv

    def calcular_tarifa(self) -> float:
        """Sobrescribe la tarifa: base 10,000 + recargo si es SUV."""
        tarifa_base = 10000.0
        return tarifa_base + 3000.0 if self.es_suv else tarifa_base


class Moto(Vehiculo):
    """Clase hija que representa una Motocicleta."""

    def __init__(self, placa: str, marca: str, cilindraje: int):
        super().__init__(placa, marca)
        self.cilindraje = cilindraje

    def calcular_tarifa(self) -> float:
        """Sobrescribe la tarifa: base 5,000 + recargo si supera 250cc."""
        tarifa_base = 5000.0
        return tarifa_base + 2000.0 if self.cilindraje > 250 else tarifa_base


if __name__ == "__main__":
    flota = [
        Auto("ABC-123", "Toyota", es_suv=True),
        Auto("XYZ-789", "Chevrolet", es_suv=False),
        Moto("MTO-456", "Yamaha", cilindraje=300),
        Moto("MTO-111", "Honda", cilindraje=125),
    ]

    print("--- REGISTRO DE PEAJE (POLIMORFISMO) ---")
    total_recaudado = 0.0

    for v in flota:
        v.mostrar_detalle()
        tarifa = v.calcular_tarifa()
        print(f"Tarifa a pagar: ${tarifa:,.2f}\n")
        total_recaudado += tarifa

    print(f"TOTAL RECAUDADO: ${total_recaudado:,.2f}")
```
