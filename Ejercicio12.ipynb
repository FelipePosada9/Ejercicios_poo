```python
# Ejercicio 12
# Autor: Felipe Posada

class Salario:
    def __init__(self, horas, pago_hora):
        self.horas = horas
        self.pago_hora = pago_hora
        self.salario_bruto = 0
        self.retencion = 0
        self.salario_neto = 0

    def calcular(self):
        self.salario_bruto = self.horas * self.pago_hora
        self.retencion = self.salario_bruto * 0.125
        self.salario_neto = self.salario_bruto - self.retencion

    def mostrar(self):
        print("SALARIO BRUTO:", self.salario_bruto)
        print("RETENCIÓN:", self.retencion)
        print("SALARIO NETO:", self.salario_neto)

obj = Salario(48, 5000)
obj.calcular()
obj.mostrar()
```
