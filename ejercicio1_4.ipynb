```
import ipywidgets as widgets
from IPython.display import display

class PruebaExcepciones:

    def ejecutar(self) -> list:
        resultados = []

        try:
            resultados.append(("Ingresando al primer try", "try"))
            cociente = 10000 / 0                        # Lanza ZeroDivisionError
            resultados.append(("Después de la división", "try"))
        except ZeroDivisionError:
            resultados.append(("División por cero", "except"))
        finally:
            resultados.append(("Ingresando al primer finally", "finally"))

        # ── Segundo bloque try ────────────────────────────────────────────
        try:
            resultados.append(("Ingresando al segundo try", "try"))
            objeto = None
            objeto.upper()                              # Lanza AttributeError
            resultados.append(("Imprimiendo objeto", "try"))
        except ZeroDivisionError:
            resultados.append(("División por cero", "except"))
        except Exception:
            resultados.append(("Ocurrió una excepción", "except"))
        finally:
            resultados.append(("Ingresando al segundo finally", "finally"))

        return resultados


# ─────────────────────────────────────────────────────────────────────────────
# Interfaz gráfica con ipywidgets (solo widgets, sin HTML de IPython)
# ─────────────────────────────────────────────────────────────────────────────

class Aplicacion:

    COLORES = {
        "try":     "#38bdf8",
        "except":  "#f87171",
        "finally": "#4ade80",
    }

    def __init__(self):
        self.prueba = PruebaExcepciones()
        self._construir_ui()

    def _construir_ui(self):

        titulo = widgets.HTML("""
        <div style="background:#2a2a3e;padding:16px 24px;border-radius:10px 10px 0 0;
                    border-bottom:2px solid #7c3aed;">
          <span style="color:#a78bfa;font-size:17px;font-weight:bold;font-family:sans-serif;">
            Ejercicio 6.4 – Excepciones
          </span><br>
          <span style="color:#94a3b8;font-size:12px;font-family:sans-serif;">
            Demuestra el comportamiento de bloques try / except / finally
          </span>
        </div>
        """)

        leyenda = widgets.HTML("""
        <div style="background:#1e1e2e;padding:10px 24px;font-family:sans-serif;font-size:13px;">
          <span style="color:#38bdf8;margin-right:20px;">● Bloque try</span>
          <span style="color:#f87171;margin-right:20px;">● Excepción capturada</span>
          <span style="color:#4ade80;">● Bloque finally</span>
        </div>
        """)

        self.area_salida = widgets.HTML(
            value='<div style="background:#0f0f1a;min-height:200px;padding:16px 24px;'
                  'color:#64748b;font-family:Courier New,monospace;font-size:13px;">'
                  'Presiona ▶ Ejecutar para ver el resultado...</div>'
        )

        btn_ejecutar = widgets.Button(
            description="▶  Ejecutar programa",
            layout=widgets.Layout(width='180px', height='36px')
        )
        btn_ejecutar.style.button_color = '#7c3aed'
        btn_ejecutar.style.font_weight = 'bold'

        btn_limpiar = widgets.Button(
            description="⟳  Limpiar",
            layout=widgets.Layout(width='120px', height='36px')
        )
        btn_limpiar.style.button_color = '#374151'

        btn_ejecutar.on_click(self._ejecutar)
        btn_limpiar.on_click(self._limpiar)

        caja_botones = widgets.HBox(
            [btn_ejecutar, btn_limpiar],
            layout=widgets.Layout(padding='12px 24px', gap='12px',
                                  background_color='#1e1e2e')
        )

        pie = widgets.HTML("""
        <div style="background:#2a2a3e;padding:8px 24px;border-radius:0 0 10px 10px;
                    font-family:sans-serif;font-size:11px;color:#64748b;">
          Ejercicios de POO con Java y UML · Leonardo Bermón Angarita · UNAL
        </div>
        """)

        self.app = widgets.VBox(
            [titulo, leyenda, self.area_salida, caja_botones, pie],
            layout=widgets.Layout(
                border='1px solid #374151',
                width='620px',
                background_color='#1e1e2e'
            )
        )

    def _ejecutar(self, _=None):
        resultados = self.prueba.ejecutar()

        lineas = ""
        for msg, tipo in resultados:
            color = self.COLORES.get(tipo, "#e2e8f0")
            lineas += (
                f'<div style="color:{color};font-family:Courier New,monospace;'
                f'font-size:14px;padding:3px 0;">  {msg}</div>'
            )

        self.area_salida.value = f"""
        <div style="background:#0f0f1a;padding:16px 24px;">
          <div style="color:#64748b;font-family:Courier New,monospace;font-size:12px;margin-bottom:10px;">
            ── Ejecución del programa ──
          </div>
          {lineas}
          <div style="color:#64748b;font-family:Courier New,monospace;font-size:12px;margin-top:10px;">
            ── Fin de la ejecución ──
          </div>
        </div>
        """

    def _limpiar(self, _=None):
        self.area_salida.value = (
            '<div style="background:#0f0f1a;min-height:200px;padding:16px 24px;'
            'color:#64748b;font-family:Courier New,monospace;font-size:13px;">'
            'Presiona ▶ Ejecutar para ver el resultado...</div>'
        )

    def mostrar(self):
        display(self.app)


# ─────────────────────────────────────────────────────────────────────────────
def main():
    app = Aplicacion()
    app.mostrar()

main()
```
