```
import math


class Notas:
    """
    Clase que representa el conjunto de notas de un estudiante.
    """

    def __init__(self):
        self.notas = []

    def ingresar_notas(self):
        """
        Solicita el ingreso de cinco notas.
        """

        print("===================================")
        print("     SISTEMA DE CÁLCULO NOTAS")
        print("===================================\n")

        for i in range(5):

            while True:

                try:
                    nota = float(input(f"Ingrese la nota {i + 1}: "))

                    if nota < 0 or nota > 5:
                        print("Error: la nota debe estar entre 0 y 5\n")
                    else:
                        self.notas.append(nota)
                        break

                except ValueError:
                    print("Error: debe ingresar un número válido\n")

    def calcular_promedio(self):
        """
        Calcula el promedio de las notas.
        """

        return sum(self.notas) / len(self.notas)

    def calcular_desviacion(self):
        """
        Calcula la desviación estándar.
        """

        promedio = self.calcular_promedio()

        suma = 0

        for nota in self.notas:
            suma += math.pow(nota - promedio, 2)

        desviacion = math.sqrt(suma / len(self.notas))

        return desviacion

    def nota_mayor(self):
        """
        Retorna la nota más alta.
        """

        return max(self.notas)

    def nota_menor(self):
        """
        Retorna la nota más baja.
        """

        return min(self.notas)

    def mostrar_resultados(self):
        """
        Muestra los resultados en pantalla.
        """

        promedio = self.calcular_promedio()
        desviacion = self.calcular_desviacion()
        mayor = self.nota_mayor()
        menor = self.nota_menor()

        print("\n===================================")
        print("           RESULTADOS")
        print("===================================")

        print(f"Promedio de notas      : {promedio:.2f}")
        print(f"Desviación estándar    : {desviacion:.2f}")
        print(f"Nota mayor             : {mayor:.2f}")
        print(f"Nota menor             : {menor:.2f}")

        print("===================================")

if __name__ == "__main__":

    estudiante = Notas()

    estudiante.ingresar_notas()

    estudiante.mostrar_resultados()
```
