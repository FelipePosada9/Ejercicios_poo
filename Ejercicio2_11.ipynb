```python
class Curso:
    """Clase que representa un Curso universitario con sobrecarga de constructor."""

    def __init__(
        self,
        nombre: str,
        codigo: str,
        creditos: int = 3,
        profesor: str = "Por asignar",
        max_estudiantes: int = 30,
    ):
        """Constructor flexible que simula sobrecarga permitiendo instanciar

        el curso con 2, 3 o todos sus atributos.
        """
        self.nombre = nombre
        self.codigo = codigo
        self.creditos = creditos
        self.profesor = profesor
        self.max_estudiantes = max_estudiantes

    def mostrar_informacion(self):
        """Muestra los datos detallados del curso."""
        print(f"\n--- Ficha del Curso: {self.nombre} ---")
        print(f"• Código: {self.codigo}")
        print(f"• Créditos: {self.creditos}")
        print(f"• Profesor: {self.profesor}")
        print(f"• Capacidad Máxima: {self.max_estudiantes} estudiantes")

if __name__ == "__main__":
    curso1 = Curso(
        nombre="Fundamentos de Programación",
        codigo="FP101"
    )
    curso1.mostrar_informacion()

    curso2 = Curso(
        nombre="Estructuras de Datos",
        codigo="ED202",
        creditos=4
    )
    curso2.mostrar_informacion()

    curso3 = Curso(
        nombre="Programación Orientada a Objetos",
        codigo="POO303",
        creditos=4,
        profesor="Dr. Leonardo Bermón",
        max_estudiantes=25,
    )
    curso3.mostrar_informacion()
```
