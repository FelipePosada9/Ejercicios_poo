```python
class Pedido:
    """Clase que representa y calcula los pedidos de un restaurante."""

    def __init__(self, cliente: str):
        self.cliente = cliente

    def calcular_pedido(
        self,
        primer_plato: str,
        costo_primer_plato: float,
        bebida: str,
        costo_bebida: float,
        segundo_plato: str = None,
        costo_segundo_plato: float = 0.0,
        postre: str = None,
        costo_postre: float = 0.0,
    ) -> float:
        """Método sobrecargado flexible que calcula el costo total del pedido

        según los platos ingresados.
        """
        total = costo_primer_plato + costo_bebida + costo_segundo_plato + costo_postre

        print(f"\n--- Resumen del Pedido de {self.cliente} ---")
        print(f"- Primer plato: {primer_plato} (${costo_primer_plato:.2f})")

        if segundo_plato:
            print(f"- Segundo plato: {segundo_plato} (${costo_segundo_plato:.2f})")

        print(f"- Bebida: {bebida} (${costo_bebida:.2f})")

        if postre:
            print(f"- Postre: {postre} (${costo_postre:.2f})")

        print(f"TOTAL A PAGAR: ${total:.2f}")
        return total


if __name__ == "__main__":
    pedido1 = Pedido("Carlos")
    pedido1.calcular_pedido(
        primer_plato="Sopa de Verduras",
        costo_primer_plato=12000.0,
        bebida="Jugo de Naranja",
        costo_bebida=5000.0,
    )
    pedido2 = Pedido("María")
    pedido2.calcular_pedido(
        primer_plato="Crema de Champiñones",
        costo_primer_plato=15000.0,
        bebida="Agua Mineral",
        costo_bebida=4000.0,
        segundo_plato="Bife de Chorizo",
        costo_segundo_plato=35000.0,
    )

    pedido3 = Pedido("Andrés")
    pedido3.calcular_pedido(
        primer_plato="Ensalada Césars",
        costo_primer_plato=14000.0,
        bebida="Vino Tinto",
        costo_bebida=18000.0,
        segundo_plato="Salmón a la Plancha",
        costo_segundo_plato=40000.0,
        postre="Tiramisú",
        costo_postre=10000.0,
    )
```
