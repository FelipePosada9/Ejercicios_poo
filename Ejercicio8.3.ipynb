```python
import math


# ==========================================
# CLASE BASE
# ==========================================

class FiguraGeometrica:
    """
    Clase base para figuras geométricas.
    """

    def calcular_volumen(self):
        pass

    def calcular_superficie(self):
        pass


# ==========================================
# CLASE CILINDRO
# ==========================================

class Cilindro(FiguraGeometrica):

    def __init__(self, radio, altura):
        self.radio = radio
        self.altura = altura

    def calcular_volumen(self):
        return math.pi * math.pow(self.radio, 2) * self.altura

    def calcular_superficie(self):
        area_lateral = 2 * math.pi * self.radio * self.altura
        area_bases = 2 * math.pi * math.pow(self.radio, 2)

        return area_lateral + area_bases


# ==========================================
# CLASE ESFERA
# ==========================================

class Esfera(FiguraGeometrica):

    def __init__(self, radio):
        self.radio = radio

    def calcular_volumen(self):
        return (4 / 3) * math.pi * math.pow(self.radio, 3)

    def calcular_superficie(self):
        return 4 * math.pi * math.pow(self.radio, 2)


# ==========================================
# CLASE PIRÁMIDE
# ==========================================

class Piramide(FiguraGeometrica):

    def __init__(self, base, altura, apotema):
        self.base = base
        self.altura = altura
        self.apotema = apotema

    def calcular_volumen(self):
        area_base = math.pow(self.base, 2)

        return (area_base * self.altura) / 3

    def calcular_superficie(self):

        area_base = math.pow(self.base, 2)

        area_lateral = 2 * self.base * self.apotema

        return area_base + area_lateral


# ==========================================
# CLASE PRINCIPAL DEL SISTEMA
# ==========================================

class SistemaFiguras:

    def mostrar_menu(self):

        print("\n======================================")
        print(" SISTEMA DE FIGURAS GEOMÉTRICAS")
        print("======================================")
        print("1. Cilindro")
        print("2. Esfera")
        print("3. Pirámide")
        print("4. Salir")
        print("======================================")

    def ejecutar(self):

        while True:

            self.mostrar_menu()

            opcion = input("Seleccione una opción: ")

            if opcion == "1":
                self.procesar_cilindro()

            elif opcion == "2":
                self.procesar_esfera()

            elif opcion == "3":
                self.procesar_piramide()

            elif opcion == "4":

                print("\nPrograma finalizado.")
                break

            else:
                print("\nOpción inválida.")

    # ======================================
    # CILINDRO
    # ======================================

    def procesar_cilindro(self):

        try:

            print("\n----- CILINDRO -----")

            radio = float(input("Ingrese el radio: "))
            altura = float(input("Ingrese la altura: "))

            cilindro = Cilindro(radio, altura)

            volumen = cilindro.calcular_volumen()
            superficie = cilindro.calcular_superficie()

            print("\nRESULTADOS")
            print(f"Volumen: {volumen:.2f}")
            print(f"Superficie: {superficie:.2f}")

        except ValueError:
            print("\nError: datos inválidos.")

    # ======================================
    # ESFERA
    # ======================================

    def procesar_esfera(self):

        try:

            print("\n----- ESFERA -----")

            radio = float(input("Ingrese el radio: "))

            esfera = Esfera(radio)

            volumen = esfera.calcular_volumen()
            superficie = esfera.calcular_superficie()

            print("\nRESULTADOS")
            print(f"Volumen: {volumen:.2f}")
            print(f"Superficie: {superficie:.2f}")

        except ValueError:
            print("\nError: datos inválidos.")

    # ======================================
    # PIRÁMIDE
    # ======================================

    def procesar_piramide(self):

        try:

            print("\n----- PIRÁMIDE -----")

            base = float(input("Ingrese la base: "))
            altura = float(input("Ingrese la altura: "))
            apotema = float(input("Ingrese el apotema: "))

            piramide = Piramide(base, altura, apotema)

            volumen = piramide.calcular_volumen()
            superficie = piramide.calcular_superficie()

            print("\nRESULTADOS")
            print(f"Volumen: {volumen:.2f}")
            print(f"Superficie: {superficie:.2f}")

        except ValueError:
            print("\nError: datos inválidos.")


# ==========================================
# PROGRAMA PRINCIPAL
# ==========================================

if __name__ == "__main__":

    sistema = SistemaFiguras()

    sistema.ejecutar()
```
