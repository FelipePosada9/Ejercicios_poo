```python
class Estudiante:
    """Clase base que representa un estudiante general."""

    def __init__(self, cedula: str, nombre: str, numero_creditos: int):
        self.cedula = cedula
        self.nombre = nombre
        self.numero_creditos = numero_creditos

    def calcular_matricula(self) -> float:
        """Método polimórfico a ser sobrescrito por las subclases."""
        return 0.0

    def mostrar_datos(self):
        print(f"Estudiante: {self.nombre} | Cédula: {self.cedula}")


class EstudiantePregrado(Estudiante):
    """Subclase para estudiantes de pregrado."""

    VALOR_CREDITO = 100000.0
    BIENESTAR = 50000.0

    def __init__(self, cedula: str, nombre: str, numero_creditos: int):
        super().__init__(cedula, nombre, numero_creditos)

    def calcular_matricula(self) -> float:
        """Calcula costo: (Créditos * Valor) + Bienestar."""
        return (self.numero_creditos * self.VALOR_CREDITO) + self.BIENESTAR


class EstudiantePosgrado(Estudiante):
    """Subclase para estudiantes de posgrado."""

    def __init__(
        self,
        cedula: str,
        nombre: str,
        numero_creditos: int,
        tipo_programa: str,
        promedio: float,
    ):
        super().__init__(cedula, nombre, numero_creditos)
        self.tipo_programa = tipo_programa
        self.promedio = promedio

    def _obtener_valor_credito(self) -> float:
        """Devuelve el costo del crédito según el nivel de posgrado."""
        programa = self.tipo_programa.lower()
        if programa == "especializacion":
            return 200000.0
        elif programa == "maestria":
            return 300000.0
        elif programa == "doctorado":
            return 400000.0
        return 200000.0

    def calcular_matricula(self) -> float:
        """Calcula el total aplicando un 20% de descuento si el promedio >= 4.5."""
        valor_credito = self._obtener_valor_credito()
        subtotal = self.numero_creditos * valor_credito

        if self.promedio >= 4.5:
            subtotal *= 0.80

        return subtotal


if __name__ == "__main__":
    estudiantes = [
        EstudiantePregrado("1010", "Laura Gómez", 16),
        EstudiantePosgrado(
            "2020", "Carlos Pérez", 10, "Especialización", promedio=4.2
        ),
        EstudiantePosgrado("3030", "Ana Martínez", 12, "Maestría", promedio=4.8),
    ]

    print("--- LIQUIDACIÓN DE MATRÍCULAS (MÉTODOS POLIMÓRFICOS) ---")
    for e in estudiantes:
        e.mostrar_datos()
        costo = e.calcular_matricula()
        print(f"Costo Total Matrícula: ${costo:,.2f}\n")
```
