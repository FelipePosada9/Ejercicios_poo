```python
# Ejercicio 14
# Autor: Felipe Posada

class Potencias:
    def __init__(self, numero):
        self.numero = numero
        self.cuadrado = 0
        self.cubo = 0

    def calcular(self):
        self.cuadrado = self.numero ** 2
        self.cubo = self.numero ** 3

    def mostrar(self):
        print("Número:", self.numero)
        print("Cuadrado:", self.cuadrado)
        print("Cubo:", self.cubo)

obj = Potencias(3)
obj.calcular()
obj.mostrar()
```
